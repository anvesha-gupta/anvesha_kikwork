```python
# stopword removal 
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

words = ["nlp","is","very","interesting"]

filtered = [w for w in words if w not in stop_words]

print(filtered)
```

```python
# stemming
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["running","runs","runner"]

for w in words:
    print(stemmer.stem(w))
```

```python
# lemmatization 
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("running", pos='v'))
```

```python
# BOW
from sklearn.feature_extraction.text import CountVectorizer

text = [
    "I love NLP",
    "I love machine learning"
]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(text)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
# TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(text)

print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
# Word2Vec
!pip install gensim

sentences = [
    ["i","love","nlp"],
    ["nlp","is","fun"],
    ["i","love","machine","learning"]
]
```

```python
# CBOW
from gensim.models import Word2Vec

cbow_model = Word2Vec(
    sentences,
    vector_size=50,
    window=2, # no. of neighbouring words on each side 
    min_count=1, # min freq for a word to be counted 
    sg=0 # cbow's tag
)

print(cbow_model.wv["nlp"])
```

```python
# Skipgram
skip_model = Word2Vec(
    sentences,
    vector_size=50,
    window=2,
    min_count=1,
    sg=1
)

print(skip_model.wv["nlp"])
```

```python
# text classification (using texts dataset)
```

```python
# convert text to features 
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(texts)
```

```python
# train test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    labels,
    test_size=0.3,
    random_state=42
)
```

```python
from sklearn.naive_bayes import MultinomialNB

model = MultinomialNB() # Multinomial Naive Bayes (algo for classification)

model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)

print(predictions)
```

```python
from sklearn.metrics import accuracy_score

print("Accuracy:",
      accuracy_score(y_test, predictions))
```
